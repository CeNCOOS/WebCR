```python
# Test to make sure all stations can be found in the dataframe(s):
import json

json_path = "JSON-inputs/marine_sites_cinms.json"   # <-- update this path

with open(json_path, "r") as f:
    config = json.load(f)

station_list = config["stationNames"]

df_column_name = "locality"    
stations_in_df = df[df_column_name].unique().tolist()

present = []
missing = []

for station in station_list:
    if station in stations_in_df:
        present.append(station)
    else:
        missing.append(station)


print("\nStations FOUND in DataFrame:")
for st in present:
    print("  ✓", st)

print("\nStations MISSING from DataFrame:")
for st in missing:
    print("  ✗", st)
```
